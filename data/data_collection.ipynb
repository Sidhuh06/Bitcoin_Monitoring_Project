```python
import requests
import pandas as pd
import sqlite3

# API URL
url = "https://api.coingecko.com/api/v3/coins/bitcoin/market_chart"

# Parameters
params = {
    "vs_currency": "usd",
    "days": "365"
}

# Request data
response = requests.get(url, params=params)

# Convert JSON
data = response.json()

# Extract prices
prices = data["prices"]

# Create DataFrame
df = pd.DataFrame(prices, columns=["timestamp", "price"])

# Convert timestamp
df["timestamp"] = pd.to_datetime(df["timestamp"], unit="ms")

# Connect to SQLite database
conn = sqlite3.connect("bitcoin_prices.db")

# Save DataFrame to SQLite table
df.to_sql("bitcoin_prices", conn, if_exists="replace", index=False)

# Verify data
query = "SELECT * FROM bitcoin_prices LIMIT 5;"
result = pd.read_sql(query, conn)

print(result)

# Close connection
conn.close()

print("Data saved successfully to bitcoin_prices.db")
```

```python
import sqlite3
import pandas as pd

# Connect to SQLite database
conn = sqlite3.connect("bitcoin_prices.db")

# SQL Query
query = """
SELECT *
FROM bitcoin_prices;
"""

# Read data into DataFrame
df = pd.read_sql(query, conn)

# Save DataFrame as CSV
df.to_csv("bitcoin_prices.csv", index=False)

# Display first rows
print(df.head())

# Close connection
conn.close()

print("CSV file saved successfully as bitcoin_prices.csv")
```
